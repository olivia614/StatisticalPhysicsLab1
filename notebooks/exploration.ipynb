# Lab 1 - interactive exploration

Quick checks of the `src/` package. The report figures are produced by `python main.py`; this notebook is for exploring other values of N.

```python
import os, sys
os.environ.setdefault("MKL_THREADING_LAYER", "SEQUENTIAL")
sys.path.insert(0, os.path.abspath(".."))
import numpy as np
import matplotlib.pyplot as plt
from src import microstates as ms, combinatorics as cb, thermodynamics as th
```

```python
# Enumeration vs. exact binomial for small N
for n in (5, 7, 10):
    print(n, list(ms.count_macrostates_enumeration(n)), cb.multiplicity_row(n))
```

```python
# Exact distribution vs. Gaussian approximation
n = 117
k = np.arange(n + 1)
plt.plot(k, th.prob(n, k), '.', label='exact')
plt.plot(k, th.gaussian_approx(n, k), '-', label='Gaussian')
plt.xlabel('k'); plt.ylabel('P_k'); plt.legend(); plt.show()
```
